```python
from google.colab import userdata
GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')
```

```python
import os
from google import genai
from google.genai import types
```

```python
messages = []
```

```python
client = genai.Client(
        api_key=GEMINI_API_KEY,
    )

generate_content_config = types.GenerateContentConfig(
        temperature=0.5,
        max_output_tokens=1024,
        system_instruction="Your are a math tutor. solve answer step by step explaination",
        thinking_config=types.ThinkingConfig(
            thinking_level="HIGH",
        ),
    )
```

```python
model = "gemini-flash-latest"
```

```python
def add_user_message(messages,text):
  messages.append(
                types.Content(
                    role="user",
                    parts=[types.Part.from_text(text=text)],
                )
            )

def add_assistant_message(messages,text):
  messages.append(
                types.Content(
                    role="model",
                    parts=[types.Part.from_text(text=text)],
                )
            )
```

```python
def generate(messages):
    full_response = []
    for chunk in client.models.generate_content_stream(
        model=model,
        contents=messages,
        config=generate_content_config
    ):
        if text := chunk.text:
            print(text, end="")
            full_response.append(text)
    return "".join(full_response)
```

```python
add_assistant_message(messages,"""At its simplest level, a **computer** is an electronic device that manipulates information, or "data." It has the ability to store, retrieve, and process data.""")
```

```python
def chat():
  messages = []

  while True:
    print("--"*50)
    print("You: ")
    print("--"*50)
    user_input = input("")
    add_user_message(messages,user_input)
    print("--"*50)
    print("Gemini: ")
    print("--"*50)
    output = generate(messages)
    print("\n")
    print("--"*50)
    add_assistant_message(messages,output)

chat()
```

```python
OPENROUTER_API_KEY = userdata.get('OPENROUTER_API_KEY')
```

```python
messages = []
```

```python
import requests
import json

def generate2(messages):
  messages += [{
      "role": "system",
      "content": "Your are a math tutor. solve answer step by step explaination"
    }] + messages

  response = requests.post(
    url="https://openrouter.ai/api/v1/chat/completions",
    headers={
      "Authorization": f"Bearer {OPENROUTER_API_KEY}",
      "Content-Type": "application/json"},
    data=json.dumps({
      "model": "openrouter/free",
      "messages": messages,
      "stream": True
    }),
    stream=True # Crucial: tells requests to stream the response
  )

  full_response_content = ""
  for line in response.iter_lines():
    if line:
      decoded_line = line.decode('utf-8')
      if decoded_line.startswith('data:'):
        json_str = decoded_line[len('data:'):].strip()
        if json_str == '[DONE]':
          break
        try:
          chunk = json.loads(json_str)
          if 'choices' in chunk and len(chunk['choices']) > 0:
            delta = chunk['choices'][0]['delta']
            if 'content' in delta:
              content_part = delta['content']
              print(content_part, end="") # Print as it arrives
              full_response_content += content_part
        except json.JSONDecodeError:
          # This might happen for non-JSON 'data:' lines or malformed JSON
          pass
  return full_response_content
```

```python
def add_user_message2(messages,text):
  messages.append({"role":"user","content":text})

def add_assistant_message2(messages,text):
  messages.append({"role":"model","content":text})
```

```python
add_user_message2(messages,"what is ai?")
```

```python
def chat2():
  messages = []

  while True:
    print("--"*50)
    print("You: ")
    print("--"*50)
    user_input = input("")
    add_user_message2(messages, user_input) # Add user input to global messages
    print("--"*50)
    print("AI: ")
    print("--"*50)
    output_content = generate2(messages) # Get text content from generate2
    print("\n") # New line after the streamed output
    print("--"*50)
    add_assistant_message2(messages, output_content) # Add assistant response to global messages
```
